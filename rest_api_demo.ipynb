# Демонстрация REST API: поиск фильма через OMDb API

Практическое занятие 1. Курс «Введение в профессию».

Пример отправки HTTP-запроса к OMDb API и разбора ответа в формате JSON.

## 1. Подключаем библиотеки

- `requests` — отправка HTTP-запросов;
- `json` — работа с JSON;
- `getpass` — безопасный ввод API-ключа без показа на экране.

```python
import json
import getpass
import requests
```

## 2. Вводим название фильма и API-ключ

Пользователь сам вводит название любимого фильма. Ключ API вводится вручную и не хранится в файле, поэтому не попадает в репозиторий.

```python
title = input("Введите фильм для поиска: ")
api_key = getpass.getpass("Введите ваш OMDb API key: ")
```

## 3. Отправляем запрос к OMDb API

GET-запрос с параметрами: `t` — название фильма, `apikey` — ключ доступа. `raise_for_status()` сообщит об ошибке, если запрос неуспешен.

```python
response = requests.get(
    "http://www.omdbapi.com/",
    params={
        "t": title,
        "apikey": api_key
    }
)
response.raise_for_status()
```

## 4. Превращаем ответ в JSON

`response.json()` превращает ответ сервиса в объект Python, а `json.dumps(..., indent=2)` печатает JSON удобно для чтения. На этом шаге видно, какие поля вообще пришли в ответе.

```python
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

## 5. Достаём нужные поля

После преобразования JSON ведёт себя как словарь Python. Метод `.get()` безопаснее обращения по квадратным скобкам.

```python
movie_title = json_data.get("Title")
year = json_data.get("Year")
genre = json_data.get("Genre")
director = json_data.get("Director")
imdb_rating = json_data.get("imdbRating")

print(f"Название: {movie_title}")
print(f"Год: {year}")
print(f"Жанр: {genre}")
print(f"Режиссёр: {director}")
print(f"Рейтинг IMDb: {imdb_rating}")
```
