# Демонстрация REST API: вариант с .env

В этом варианте API-ключ хранится в файле `.env` и читается через библиотеку `python-dotenv`. Файл `.env` добавлен в `.gitignore` и в репозиторий не попадает.

Эта ветка (`env-feature`) в `master` не вливается.

## 1. Подключаем библиотеки

- `os` — доступ к переменным окружения;
- `json` — работа с JSON;
- `requests` — отправка HTTP-запросов;
- `dotenv.load_dotenv` — загрузка переменных из файла `.env`.

```python
import os
import json
import requests
from dotenv import load_dotenv

load_dotenv()
```

## 2. Вводим название фильма и читаем ключ из .env

Ключ больше не вводится с клавиатуры — он берётся из переменной окружения `OMDB_API_KEY`.

```python
title = input("Введите фильм для поиска: ")
api_key = os.getenv("OMDB_API_KEY")

if not api_key:
    raise ValueError("Ключ OMDB_API_KEY не найден в .env")
```

## 3. Отправляем запрос к OMDb API

```python
response = requests.get(
    "http://www.omdbapi.com/",
    params={
        "t": title,
        "apikey": api_key
    }
)
response.raise_for_status()
```

## 4. Превращаем ответ в JSON и проверяем результат

По совету code review добавили проверку поля `Response`: если фильм не найден, выводим понятное сообщение, а не «сломанный» сценарий.

```python
json_data = response.json()

if json_data.get("Response") == "False":
    print("Фильм не найден:", json_data.get("Error"))
else:
    print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

## 5. Достаём нужные поля

```python
movie_title = json_data.get("Title")
year = json_data.get("Year")
genre = json_data.get("Genre")
director = json_data.get("Director")
imdb_rating = json_data.get("imdbRating")

print(f"Название: {movie_title}")
print(f"Год: {year}")
print(f"Жанр: {genre}")
print(f"Режиссёр: {director}")
print(f"Рейтинг IMDb: {imdb_rating}")
```
